# Dicionario de Dados - Mapa de Salas

Este dicionario documenta o arquivo `mapa_salas_tidy_01.csv`, localizado na mesma pasta desta atividade. A base foi gerada a partir dos arquivos originais tratados pelo projeto.

**Unidade de observacao:** cada linha representa um encontro de uma turma de um curso em uma sala, em um dia da semana e intervalo continuo de horarios. Uma turma que atende mais de um curso aparece uma vez para cada curso.

**Cobertura:** periodo letivo 2026/2, 23 salas da Faculdade de Arquitetura. O arquivo possui 440 registros e 16 atributos; nao ha valores ausentes.

## Atributos

| Atributo | Significado | Tipo computacional | Tipo semantico | Exemplo | Observacoes |
|---|---|---|---|---|---|
| `semestre` | Periodo letivo da oferta. | texto | temporal/categorico | `2026/2` | O arquivo atual contem apenas um semestre. |
| `predio` | Predio que contem a sala. | texto | categorico/geografico | `Faculdade de Arquitetura - 12103` | Permite agregar ocupacao por predio. |
| `sala` | Identificacao da sala onde ocorre o encontro. | texto | identificador/categorico | `Sala 103` | Extraida do cabecalho de cada aba do arquivo original. |
| `tipo_sala` | Tipo ou finalidade predominante da sala. | texto | categorico | `Sala de aula`, `Atelier`, `Laboratorio de Informatica` | Extraido do cabecalho de cada aba; permite comparar ocupacao por infraestrutura. |
| `capacidade_turma` | Quantidade maxima de estudantes comportada pela sala. | inteiro | numerico discreto | `16` | Extraida do cabecalho de cada aba; nao confundir com vagas oferecidas da turma. |
| `codigo_disciplina` | Codigo institucional da disciplina. | texto | identificador | `ARQ01045` | Nao deve ser convertido para numero. |
| `turma` | Identificador da turma da disciplina. | texto | categorico | `A`, `AA`, `C` | Turmas que apareciam agregadas com `/` foram separadas. |
| `nome_disciplina` | Nome da atividade ou disciplina. | texto | categorico | `MAQUETES` | Associado a grade pelo identificador de turma original. |
| `curso` | Curso atendido pela turma. | texto | categorico | `ARQU`, `DVIS` | Uma turma de varios cursos gera um registro por curso. |
| `dia_semana` | Dia em que o encontro ocorre. | texto | temporal/categorico ordenado | `SEGUNDA-FEIRA` | Derivado do cabecalho das colunas da grade semanal. |
| `hora_inicio` | Horario de inicio do encontro. | texto no formato `HH:MM` | temporal | `09:30` | Primeiro periodo de uma sequencia continua. |
| `hora_fim` | Horario de termino do encontro. | texto no formato `HH:MM` | temporal | `12:30` | Corresponde ao fim do ultimo periodo da sequencia. |
| `numero_periodos` | Quantidade de periodos consecutivos de uma hora. | inteiro | numerico discreto | `3` | Usado para calcular duracao e ocupacao. |
| `vagas_oferecidas` | Vagas atribuidas a uma turma individual. | inteiro | numerico discreto | `18` | Em turmas compartilhadas, as vagas totais foram divididas; a primeira turma recebeu eventual resto. |
| `vagas_totais_compartilhadas` | Total de vagas informado para o grupo de turmas que compartilha o encontro. | inteiro | numerico discreto | `36` | Evita perda de informacao da planilha original. Nao somar entre linhas de uma mesma turma compartilhada. |
| `turmas_compartilhando_sala` | Conjunto de turmas que usa simultaneamente a mesma sala. | texto | categorico | `A/B`, `A/AA`, `C/D` | Use para identificar compartilhamento de espaco. |
